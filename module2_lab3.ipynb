### Temperature Classification Rules
* **Freezing:** $T < 32^\circ\text{F}$
* **Cold:** $32^\circ\text{F} \le T < 50^\circ\text{F}$
* **Mild:** $50^\circ\text{F} \le T < 70^\circ\text{F}$
* **Warm:** $70^\circ\text{F} \le T < 85^\circ\text{F}$
* **Hot:** $T \ge 85^\circ\text{F}$

```python
temperatures = [25, 32, 45, 50, 65, 70, 78, 85, 95]

for temp in temperatures:
    if temp < 32:
        category = "Freezing"
    elif temp < 50:
        category = "Cold"
    elif temp < 70:
        category = "Mild"
    elif temp < 85:
        category = "Warm"
    else:
        category = "Hot"
    
    print(f"{temp}°F: {category}")
```
